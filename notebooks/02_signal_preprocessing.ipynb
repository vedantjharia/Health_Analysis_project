# Notebook 02: Biosignal Filtering & R-Peak Detection

Demonstration of 4th-order Butterworth bandpass filtering (0.5-45 Hz), Pan-Tompkins R-peak detection, and respiratory rhythm extraction.


```python
import sys
sys.path.append('..')

import numpy as np
import matplotlib.pyplot as plt
from src.data_loader import FantasiaDataLoader
from src.signal_processing import BiosignalProcessor
```

```python
loader = FantasiaDataLoader()
processor = BiosignalProcessor()

sub = loader.fetch_subject_signal('f1y01')
ecg_filt = processor.filter_ecg(sub['ecg'])
resp_filt = processor.filter_respiration(sub['resp'])
r_peaks = processor.detect_r_peaks(ecg_filt)
rr_times, rr_ms = processor.extract_rr_intervals(r_peaks)

print(f'Detected {len(r_peaks)} R-peaks. Mean RR: {np.mean(rr_ms):.1f} ms')
```

```python
plt.figure(figsize=(12, 4))
plt.plot(ecg_filt[:2500], color='#1b4f72', label='Filtered ECG (0.5-45 Hz)')
p_mask = r_peaks < 2500
plt.scatter(r_peaks[p_mask], ecg_filt[r_peaks[p_mask]], color='#c0392b', s=50, label='Pan-Tompkins R-peaks')
plt.legend()
plt.title('Pan-Tompkins Peak Detection on Cleaned ECG')
plt.show()
```
