# Notebook 03: Subject-Level Holdout Training & Evaluation

Trains benchmark models with strict Subject-Level Holdout partitioning to eliminate signal window data leakage.


```python
import sys
sys.path.append('..')

import pandas as pd
from src.train import load_dataset, split_by_subject, build_models, train_and_cross_validate
from src.utils import load_config
```

```python
config = load_config()
df = load_dataset()
test_subjects = config['evaluation']['test_subjects']
train_df, test_df = split_by_subject(df, test_subjects)

print(f'Training windows: {len(train_df)}, Testing windows: {len(test_df)}')
```

```python
models = build_models(config)
cv_res = train_and_cross_validate(train_df, models)

pd.DataFrame(cv_res).T[['cv_accuracy_mean', 'cv_f1_mean', 'cv_roc_auc_mean']]
```
